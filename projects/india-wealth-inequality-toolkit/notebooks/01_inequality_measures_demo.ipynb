# Inequality Measures — Demonstration

**This notebook uses a SYNTHETIC dataset** (`data/synthetic_household_wealth.csv`), generated by `examples/generate_synthetic_wealth.py`. It is not real survey data. Every number below describes the synthetic generator, not India — the point is to demonstrate the toolkit's API, not to report an empirical finding.

```python
import sys
sys.path.insert(0, '../src')
import pandas as pd
from india_inequality import inequality as ineq

df = pd.read_csv('../data/synthetic_household_wealth.csv')
values = df['net_worth_inr_lakhs']
df.describe()
```

```python
print('Gini:', ineq.calculate_gini(values))
print('Top 10% share:', ineq.top_decile_share(values))
print('Bottom 50% share:', ineq.bottom_half_share(values))
print('Palma ratio:', ineq.palma_ratio(values))
```

```python
import matplotlib.pyplot as plt
curve = ineq.lorenz_curve(values)
plt.plot(curve['population_share'], curve['value_share'], label='Lorenz curve')
plt.plot([0, 1], [0, 1], '--', color='gray', label='Perfect equality')
plt.xlabel('Cumulative population share')
plt.ylabel('Cumulative value share')
plt.legend()
plt.title('Lorenz curve — SYNTHETIC data')
plt.show()
```

## Composition effect (wealth module)

Deterministic comparison of two fixed asset allocations under identical contributions. See `docs/METHODOLOGY.md` for the exact update rule and assumptions.

```python
from india_inequality import wealth

property_heavy = wealth.composition_effect(100_000, 20, wealth.DEFAULT_ALLOCATION, inflation=0.05)
equity_heavy = wealth.composition_effect(
    100_000, 20,
    {'property': 0.10, 'gold': 0.05, 'equities': 0.65, 'gov_bonds': 0.10, 'bank_deposits': 0.08, 'cash': 0.02},
    inflation=0.05,
)
print('Property-heavy real wealth after 20y:', round(property_heavy.real_wealth[-1]))
print('Equity-heavy real wealth after 20y:', round(equity_heavy.real_wealth[-1]))
```
